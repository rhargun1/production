```python
import os
from glob import glob

PRICE_DATA = os.getenv("PRICE_DATA")

if PRICE_DATA is None:
    raise ValueError("PRICE_DATA is not set. Check your .env file or environment.")

parquet_files = glob(os.path.join(PRICE_DATA, "**", "*.parquet"), recursive=True)

if not parquet_files:
    raise FileNotFoundError(f"No Parquet files found under {PRICE_DATA}")
```

For each ticker and using Dask, do the following:

+ Add lags for variables Close and Adj_Close.
+ Add returns based on Close:
    
    - `returns`: (Close / Close_lag_1) - 1

+ Add the following range: 

    - `hi_lo_range`: this is the day's High minus Low.

+ Assign the result to `dd_feat`.

(4 pt)

```python
# Write your code below.
# Read parquet files using Dask
dd_feat = dd.read_parquet(parquet_files)

# Ensure observations are ordered by ticker and date
dd_feat = dd_feat.map_partitions(
    lambda df: df.sort_values(["ticker", "Date"])
)

# Lag features
dd_feat["Close_lag_1"] = (
    dd_feat.groupby("ticker")["Close"].shift(1)
)

dd_feat["Adj_Close_lag_1"] = (
    dd_feat.groupby("ticker")["Adj Close"].shift(1)
)

# Daily returns
dd_feat["returns"] = (
    dd_feat["Close"] / dd_feat["Close_lag_1"]
) - 1

# Daily high-low range
dd_feat["hi_lo_range"] = (
    dd_feat["High"] - dd_feat["Low"]
)

dd_feat.columns
```

+ Convert the Dask data frame to a pandas data frame. 
+ Add a new feature containing the moving average of `returns` using a window of 10 days. There are several ways to solve this task, a simple one uses `.rolling(10).mean()`.

(3 pt)

```python
# Write your code below.
# Convert Dask DataFrame to pandas
df_feat = dd_feat.compute()
```

```python
# Write your code below.
# Convert Dask DataFrame to pandas
df_feat = dd_feat.compute()

# Sort observations before applying rolling window
df_feat = df_feat.sort_values(["ticker", "Date"])

# 10-day moving average of returns for each ticker
df_feat["returns_ma_10"] = (
    df_feat.groupby("ticker")["returns"]
    .rolling(10)
    .mean()
    .reset_index(level=0, drop=True)
)

##
print(PRICE_DATA)
print(len(parquet_files))
print(df_feat.columns)
```
